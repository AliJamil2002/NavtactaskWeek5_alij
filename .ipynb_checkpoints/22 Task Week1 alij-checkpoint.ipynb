```python

my_tuple = ()
print(my_tuple)
# tuple having integers
my_tuple = (1, 2, 3)
print(my_tuple)
# tuple with mixed datatypes
my_tuple = (1, "Hello", 3.4)
print(my_tuple)
# nested tuple
my_tuple = ("mouse", [8, 4, 6], (1, 2, 3))
print(my_tuple)
# tuple can be created without parentheses
# also called tuple packing
my_tuple = 3, 4.6, "dog"
print(my_tuple)
# tuple unpacking is also possible
a, b, c = my_tuple
print(my_tuple)
```

```python

my_tuple = ['p','e','r','m','i','t']
print(my_tuple[0])

print(my_tuple[5])

print(my_tuple[4])   # index must be in range


print(my_tuple[2.0]) # index must be an integer



 
```

```python
n_tuple = ("mouse", [8, 4, 6], (1, 2, 3))
n_tuple[0][3]  # nested index

n_tuple[1][1]  # nested index

n_tuple[2][0]  # nested index
```

```python

my_tuple = ['p','e','r','m','i','t']
print(my_tuple[-1])
print(my_tuple[-6])
print(my_tuple[:-3])
```

```python

my_tuple = ('p','r','o','g','r','a','m','i','z')
print(my_tuple[1:4])

print(my_tuple[:-7])

print(my_tuple[7:])  

print(my_tuple[:])   
```
