```python
import numpy as np

a = np.array([1, 2, 4, 5, 7, 8])
b = np.array([7, 6, 8, 9, 8])

ans = np.concatenate((a, b))

print(ans)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
ans = np.concatenate((a, b), axis=0)
print(ans)
```
